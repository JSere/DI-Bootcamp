## Exercise 1: Defining the Problem and Data Collection for Loan Default Prediction

### Problem Statement
- **Objective**: Build a predictive classification model to determine whether a loan applicant is likely to default on their loan repayment.
- **Business Impact**: Minimize financial risk and reduce non-performing loans (NPLs) by identifying high-risk applicants during the underwriting stage, while maintaining a fair approval rate for creditworthy candidates.

### Types of Data Needed
1. **Demographic & Personal Details**: Age, marital status, education level, housing status (rent/own), and number of dependents.
2. **Financial Background**: Annual income, employment status, length of employment, and debt-to-income (DTI) ratio.
3. **Credit History & Bureau Data**: Credit score (FICO), number of open credit lines, history of past delinquencies, bankruptcies, and credit utilization rate.
4. **Loan-Specific Details**: Requested loan amount, loan term (36 or 60 months), purpose of the loan (debt consolidation, home improvement), and proposed interest rate.

### Potential Data Sources
- **Internal Institutional Records**: Past historical loan performance data, customer banking history, and internal transaction patterns.
- **External Credit Bureaus**: Consumer credit reports from major bureaus (e.g., Equifax, Experian, TransUnion).
- **Alternative Data Providers**: Employment verification services, tax assessment records, and utility/rent payment history for thin-file applicants.

## Exercise 2: Feature Selection and Model Choice for Loan Default Prediction

### Selected Features & Rationale
1. **Debt-to-Income (DTI) Ratio**: Measures an applicant's monthly debt payments against their monthly gross income. High DTI indicates stretched financial capacity, making it a powerful predictor of default.
2. **Credit Score**: Summarizes past credit behavior. A low credit score historically correlates strongly with a higher probability of default.
3. **Employment Length & Status**: Stable income ensures liquidity. Shorter employment history or unemployment increases default risk.
4. **Loan Amount & Term**: Larger loan amounts relative to income increase the absolute burden of repayment, especially over longer horizons.
5. **Recent Delinquencies / Public Records**: Past defaults or bankruptcies are primary behavioral indicators of future default risk.

## Exercise 3: Training, Evaluating, and Optimizing the Model

### Model Selection
- **Recommended Model**: **XGBoost** or **Random Forest Classifier**.
- **Justification**: Loan default datasets are tabular, heterogeneous, and contain non-linear relationships. Tree-based ensemble methods handle missing values effectively, capture complex feature interactions, and resist overfitting.

### Evaluation Strategy & Metrics
- **Precision-Recall AUC (PR-AUC)**: Preferred over ROC-AUC when dealing with class imbalance (since defaults are usually a minority class).
- **F1-Score**: Balances precision (minimizing false positives, i.e., denying good customers) and recall (minimizing false negatives, i.e., approving defaults).
- **Stratified K-Fold Cross-Validation**: Ensures each fold maintains the same proportion of default vs. non-default cases.

## Exercise 5: Designing an Evaluation Strategy for Different ML Models

1. **Supervised Learning (Classification Model)**
   - **Metrics**: Accuracy, Precision, Recall, F1-Score, and ROC-AUC.
   - **Methods**: $k$-fold Cross-Validation and holdout test sets.
   - **Challenges & Limitations**: Highly sensitive to class imbalance, label noise, and data leakage during feature engineering.

2. **Unsupervised Learning (Clustering Model)**
   - **Metrics**: Silhouette Score, Calinski-Harabasz Index, and the Elbow Method (Inertia).
   - **Methods**: Internal validation metrics measuring cluster cohesion (compactness) and separation.
   - **Challenges & Limitations**: Lack of ground-truth labels makes objective performance evaluation difficult; optimal cluster numbers ($k$) are often subjective or domain-dependent.

3. **Reinforcement Learning Model**
   - **Metrics**: Cumulative reward per episode, convergence rate (speed to optimal policy), and success rate across varying environments.
   - **Methods**: Tracking policy performance over training epochs and evaluating against baseline heuristic agents.
   - **Challenges & Limitations**: Sample inefficiency (requiring vast amounts of training episodes), the credit assignment problem (determining which action led to a delayed reward), and balancing exploration versus exploitation.
