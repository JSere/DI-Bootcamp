```python
# Install statsmodels if running in Google Colab or a fresh environment
!pip install -q statsmodels

import numpy as np
from statsmodels.stats.power import NormalIndPower

# --- Task 1 & 2: Calculate Required Sample Size & Effect Size Impact ---
alpha = 0.05
power = 0.80
effect_sizes = [0.1, 0.2, 0.3, 0.4]

power_analysis = NormalIndPower()

print("=== Sweet Bytes A/B Test Sample Size Analysis ===")
for es in effect_sizes:
  sample_size = power_analysis.solve_power(
      effect_size=es, alpha=alpha, power=power, ratio=1.0
  )
  print(f"Effect Size {es}: {np.ceil(sample_size)} customers per group")
```
