```python
!pip install -q statsmodels

import numpy as np
from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize

# --- Exercise 1: Calculating Required Sample Size ---
# Effect size based on open rates increasing from 20% to 23%
effect_size_ex1 = proportion_effectsize(0.23, 0.20)
alpha = 0.05
power = 0.80

power_analysis = NormalIndPower()
sample_size_ex1 = power_analysis.solve_power(
    effect_size=effect_size_ex1, alpha=alpha, power=power, ratio=1.0
)
print(f"Exercise 1 - Required sample size per group: {np.ceil(sample_size_ex1)}")

# --- Exercise 2: Understanding the Relationship Between Effect Size and Sample Size ---
effect_sizes = [0.2, 0.4, 0.5]
print("\nExercise 2 - Sample size for different effect sizes:")
for es in effect_sizes:
  n = power_analysis.solve_power(
      effect_size=es, alpha=alpha, power=power, ratio=1.0
  )
  print(f"Effect size {es}: {np.ceil(n)} per group")

# --- Exercise 3: Exploring the Impact of Statistical Power ---
power_levels = [0.7, 0.8, 0.9]
base_es = 0.2
print("\nExercise 3 - Sample size for different power levels:")
for p in power_levels:
  n = power_analysis.solve_power(
      effect_size=base_es, alpha=alpha, power=p, ratio=1.0
  )
  print(f"Power {p}: {np.ceil(n)} per group")
```
