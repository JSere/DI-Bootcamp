```python
# Original Policy Input Text
input_text = """Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."""

# --- Step 1: Craft a Prompt That Controls Output Tone, Format, and Length ---
step1_prompt = f"""
Rewrite the following policy text in a friendly and clear tone. 
Use bullet points to organize the information. 
Paraphrase the original content without using direct quotes. 
Keep the total response under 75 words.

Input Text:
{input_text}
"""
print("=== Step 1 Prompt ===")
print(step1_prompt.strip())
print("\n" + "="*50 + "\n")


# --- Step 2: Evaluation Framework ---
evaluation_criteria = {
    "Relevance": "Does the output stick to the message of the source content?",
    "Clarity": "Is the message easier to understand for non-technical staff?",
    "Structure": "Is the information formatted into clear bullet points?",
    "Tone": "Is the language appropriate for internal employee communication?",
    "Length": "Does it stay under 75 words?",
    "Factual Accuracy": "Did the model make up or remove any critical security details?"
}
print("=== Step 2: Evaluation Criteria Defined ===")
for criterion, question in evaluation_criteria.items():
    print(f"- {criterion}: {question}")
print("\n" + "="*50 + "\n")


# --- Step 3: Detect and Mitigate Hallucinations ---
step3_prompt = f"""
Rewrite the following policy text in a friendly and clear tone using bullet points. 
Strict rules:
- Only paraphrase the content in the input text. 
- Do not add new recommendations, tools, or technologies not explicitly mentioned.
- If uncertain, state that the information is not available.
- Keep the total response under 75 words.

Input Text:
{input_text}
"""
print("=== Step 3 Anti-Hallucination Prompt ===")
print(step3_prompt.strip())
print("\n" + "="*50 + "\n")


# --- Step 4: Paraphrasing Deep Dive (Junior Intern Audience) ---
step4_prompt = f"""
Paraphrase the following security policy for a junior intern audience. 
Use plain language and short phrases.
Constraints:
- Use no more than 4 bullet points.
- Avoid corporate or legal jargon.
- Keep the tone supportive and informative.

Input Text:
{input_text}
"""
print("=== Step 4 Audience-Specific Prompt ===")
print(step4_prompt.strip())
print("\n" + "="*50 + "\n")


# --- Step 5: Quote Extraction Variant & Answers ---
step5_prompt = f"""
Extract one direct quote from the original text that best captures the core security policy.

Input Text:
{input_text}
"""

quote_analysis = {
    "Appropriating quoting": "Official compliance announcements, legal policy handbooks, or security audits where exact wording is mandatory.",
    "Risks of quoting": "Can confuse non-technical staff due to dense corporate jargon, leading to poor comprehension."
}

print("=== Step 5 Quote Extraction Prompt ===")
print(step5_prompt.strip())
print("\n--- Quote Analysis ---")
for key, value in quote_analysis.items():
    print(f"{key}: {value}")
```
