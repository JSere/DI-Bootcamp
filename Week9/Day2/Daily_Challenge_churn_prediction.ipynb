# Applying Inferential Statistics

### Here are the hypotheses to test:
1. Age of people who left the bank and who did not are similar. Alternative: Not similar.
2. Credit score of people who left the bank and who did not are similar. Alternative: Not similar.
3. Balance of people who left the bank and who did not are similar. Alternative: Not similar.
4. Estimated Salary of people who left the bank and who did not are similar. Alternative: Not similar.

#### The most appropriate test to analyse data here is Frequentist test.

```python
[ ]
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy.stats as stats
import seaborn as sns

matplotlib.rcParams['figure.figsize'] = (8.0, 5.0)
```

```python
matplotlib.rcParams['figure.figsize'] = (8.0, 5.0)
```

```python
[ ]
## TODO : load the csv file from this link : https://www.kaggle.com/code/vaibhagarwal/inferential-statistics/input
file_1 = "Bank Customer Churn Prediction.csv"
```

```python
[ ]
## TODO : make into a dataframe called df
df = pd.read_csv(file_1)
```

```python
[ ]
## TODO : output the first 5 lines
df.head()
```

```python
[ ]
## TODO : Create two separate DataFrames, `df_0` and `df_1`, to filter customers who have not exited (0) and customers who have exited (1), respectively
df_0 = df[df["Exited"] == 0]
df_1 = df[df["Exited"] == 1]
```

## Hypothesis 1: Age

```python
[ ]
## TODO: Plot the age distribution for customers who stayed with the bank and those who left using seaborn, with different colors for each group and a legend.
plt.figure(figsize=(8, 5))
sns.histplot(
    df_0["Age"],
    color="blue",
    label="Stayed (0)",
    kde=True,
    stat="density",
    alpha=0.5,
)
sns.histplot(
    df_1["Age"],
    color="red",
    label="Exited (1)",
    kde=True,
    stat="density",
    alpha=0.5,
)
plt.legend()
plt.title("Age Distribution by Churn Status")
plt.show()
```

```python
[ ]
## TODO: Calculate the mean and standard deviation of the age for customers who stayed with the bank.
mean_age_0 = df_0["Age"].mean()
std_age_0 = df_0["Age"].std()
print(f"Stayed - Mean: {mean_age_0:.2f}, Std: {std_age_0:.2f}")
```

```python
[ ]
## TODO: Calculate the mean and standard deviation of the age for customers who left the bank.
mean_age_1 = df_1["Age"].mean()
std_age_1 = df_1["Age"].std()
print(f"Exited - Mean: {mean_age_1:.2f}, Std: {std_age_1:.2f}")
```

```python
[ ]
## TODO: Perform a t-test to compare the ages of customers who stayed and left the bank.
t_stat, p_val = stats.ttest_ind(df_0["Age"], df_1["Age"], equal_var=False)
print(f"T-test Age -> t-stat: {t_stat:.4f}, p-value: {p_val:.4e}")
```

### Using Bootstrapping

```python
## TODO: Write a function to perform bootstrap sampling and calculate the statistic of interest.
def bs_choice(data, func, size):
    bs_s = np.empty(size)
    for i in range(size):
        bs_abc = np.random.choice(data, size=len(data))
        bs_s[i] = func(bs_abc)
    return bs_s
```

```python
[ ]
# Run bootstrap for Age of exited customers
bs_age_1 = bs_choice(df_1["Age"], np.mean, 10000)
print("Bootstrap Mean Age (Exited):", np.mean(bs_age_1))
```

```python
[ ]
# Additional bootstrap validation if needed
print("Bootstrap Std Age (Exited):", np.std(bs_age_1))
```

```python
[ ]
# Summary display for hypothesis 1
print(f"Observed mean difference in Age: {mean_age_1 - mean_age_0:.2f}")
```

### Conclusion
Do we reject the Null Hypothesis ? Why ?

## Hypothesis 2: Credit Score

```python
[ ]
## TODO: Create histograms for the CreditScore distribution of both groups (Still with bank and Left the bank).
plt.figure(figsize=(8, 5))
sns.histplot(
    df_0["CreditScore"],
    color="blue",
    label="Stayed (0)",
    kde=True,
    stat="density",
    alpha=0.5,
)
sns.histplot(
    df_1["CreditScore"],
    color="red",
    label="Exited (1)",
    kde=True,
    stat="density",
    alpha=0.5,
)
plt.legend()
plt.title("Credit Score Distribution by Churn Status")
plt.show()
```

```python
[ ]
## TODO: Perform a t-test to compare the CreditScore between the two groups (Still with bank and Left the bank).
t_stat, p_val = stats.ttest_ind(
    df_0["CreditScore"], df_1["CreditScore"], equal_var=False
)
print(f"T-test CreditScore -> t-stat: {t_stat:.4f}, p-value: {p_val:.4e}")
```

### Conclusion
Do we reject the Null Hypothesis ? Why ?
Depending on the p-value threshold, credit score often shows minor or no statistically significant difference between groups compared to age, meaning we might fail to reject the null hypothesis.

## Hypothesis 3: Balance

```python
[ ]
## TODO: Perform a t-test to compare the Balance between customers who stayed with the bank and those who left.
t_stat, p_val = stats.ttest_ind(df_0["Balance"], df_1["Balance"], equal_var=False)
print(f"T-test Balance -> t-stat: {t_stat:.4f}, p-value: {p_val:.4e}")
```

```python
[ ]
## TODO: Visualize the distribution of Balance for customers who stayed with the bank and those who left, excluding zero balances.
df_balance_nz = df[df["Balance"] > 0]
plt.figure(figsize=(8, 5))
sns.histplot(
    df_balance_nz[df_balance_nz["Exited"] == 0]["Balance"],
    color="blue",
    label="Stayed (0)",
    kde=True,
    stat="density",
    alpha=0.5,
)
sns.histplot(
    df_balance_nz[df_balance_nz["Exited"] == 1]["Balance"],
    color="red",
    label="Exited (1)",
    kde=True,
    stat="density",
    alpha=0.5,
)
plt.legend()
plt.title("Non-zero Balance Distribution by Churn Status")
plt.show()
```

```python
[ ]
## TODO: Perform a t-test to compare the Balance between customers who stayed with the bank and those who left, excluding zero balances.
bal_0_nz = df_0[df_0["Balance"] > 0]["Balance"]
bal_1_nz = df_1[df_1["Balance"] > 0]["Balance"]
t_stat, p_val = stats.ttest_ind(bal_0_nz, bal_1_nz, equal_var=False)
print(
    f"T-test Non-zero Balance -> t-stat: {t_stat:.4f}, p-value: {p_val:.4e}"
)
```

## Conclusion

Do we reject the Null Hypothesis ? Why ?
Yes, especially when excluding zero balances, account balance exhibits significant variance and distinction between churned and non-churned customers.

## Hypothesis 4: Estimated Salary

```python
[ ]
## TODO: Plot the distribution of EstimatedSalary for customers who stayed with the bank and those who left.
plt.figure(figsize=(8, 5))
sns.histplot(
    df_0["EstimatedSalary"],
    color="blue",
    label="Stayed (0)",
    kde=True,
    stat="density",
    alpha=0.5,
)
sns.histplot(
    df_1["EstimatedSalary"],
    color="red",
    label="Exited (1)",
    kde=True,
    stat="density",
    alpha=0.5,
)
plt.legend()
plt.title("Estimated Salary Distribution by Churn Status")
plt.show()
```

```python
[ ]
## TODO: Perform a t-test to compare the EstimatedSalary between customers who stayed and those who left.
t_stat, p_val = stats.ttest_ind(
    df_0["EstimatedSalary"], df_1["EstimatedSalary"], equal_var=False
)
print(
    f"T-test EstimatedSalary -> t-stat: {t_stat:.4f}, p-value: {p_val:.4e}"
)
```

### Using Bootstrapping

```python
[ ]
## TODO: Calculate the difference in means and shift the EstimatedSalary for both groups.
sal_0 = df_0["EstimatedSalary"]
sal_1 = df_1["EstimatedSalary"]
obs_diff = np.mean(sal_1) - np.mean(sal_0)
combined_sal = np.concatenate((sal_0, sal_1))
shift_sal_0 = sal_0 - np.mean(sal_0) + np.mean(combined_sal)
shift_sal_1 = sal_1 - np.mean(sal_1) + np.mean(combined_sal)
```

```python
[ ]
## TODO: Calculate the bootstrap sample means for both groups and their difference.
bs_reps_0 = bs_choice(shift_sal_0, np.mean, 10000)
bs_reps_1 = bs_choice(shift_sal_1, np.mean, 10000)
bs_diffs = bs_reps_1 - bs_reps_0
```

```python
[ ]
## TODO: Calculate the p-value based on the bootstrap distribution of the difference in means.
p_value = np.sum(np.abs(bs_diffs) >= np.abs(obs_diff)) / len(bs_diffs)
print(f"Bootstrap p-value: {p_value}")
```

### Conclusion
Do we reject the Null Hypothesis ? Why ?
Typically, estimated salary shows no significant difference between customers who left and those who stayed ($p > 0.05$), so we fail to reject the null hypothesis.
